# 02. Ground truth
Đọc caption từ `data/captions.csv` (đã bỏ caption lặp ở notebook 01), chỉ lấy split test.

```python
import json, os
import pandas as pd

ROOT = ".."
OUT = f"{ROOT}/artifacts/ground_truth"
os.makedirs(OUT, exist_ok=True)

caps = pd.read_csv(f"{ROOT}/data/captions.csv")
test = caps[caps["split"] == "test"].reset_index(drop=True)
print(len(test), "caption |", test["image_id"].nunique(), "ảnh")   # kỳ vọng: 4999 | 1000
```

```python
# caption_id → image_id đúng, vd. "test_0_0" → "test_0"
t2i = dict(zip(test["caption_id"], test["image_id"]))
print(len(t2i))   # kỳ vọng: 4999
```

```python
assert len(t2i) == test["caption_id"].nunique()     # caption_id không trùng
ex = test["image_id"].iloc[0]
print(test.loc[test["image_id"] == ex, ["caption_id", "caption", "image_id"]])
```

```python
json.dump(t2i, open(f"{OUT}/t2i.json", "w"), indent=1)
print(os.listdir(OUT))   # kỳ vọng có t2i.json
```
